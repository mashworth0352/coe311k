```python
import numpy as np
import matplotlib.pyplot as plt

k = 10.0
h = np.array([.01, .02, .03, .05, .1, .2, .3, .5])
x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0
plt.figure(figsize=(10, 6))
for j in range (h.size):
    
    x_values = np.arange(x_start, x_end + h[j], h[j])
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))
    
    y_values[0] = y_0
    v_values[0] = v_0
    
    for i in range(1, len(x_values)):
        
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]    

        y_values[i] = y_n + h[j] * v_n
        v_values[i] = v_n + h[j] * (-k * y_n)

    plt.plot(x_values, y_values, label=('Position y(x) for h =', h[j]))
    plt.plot(x_values, v_values, label=('Velocity v(x) for h =', h[j]), linestyle='--')

plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)
plt.show()

```

The plots are generally stable until x=6, where instability slowly becomes apparent. By x=9, there are wide variations in the plots. It is notable that as the value of h increases, the magnitue of the insability also increases.
