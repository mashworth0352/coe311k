```python
import numpy as np
import matplotlib.pyplot as plt
import time

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    
    return C

matrix_sizes = []
naive_times = []
numpy_times = []

n = 3

for matrix_number in range(8):
    
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)
    
    start_time = time.time()
    naive_result = naive_matrix_multiplication(A, B)
    end_time = time.time()
    naive_times.append(end_time - start_time)
    
    start_time = time.time()
    numpy_result = np.dot(A, B)
    end_time = time.time()
    numpy_times.append(end_time - start_time)
    
    matrix_sizes.append(n)
    n = n * 2

print("Matrix Sizes:", matrix_sizes)
print("Naive Times:", naive_times)
print("Numpy Times:", numpy_times)

plt.plot(matrix_sizes, naive_times, label='Naive')
plt.plot(matrix_sizes, numpy_times, label='Numpy')
plt.title('Matrix Number vs Runtime')
plt.xlabel('Matrix Size (n x n)')
plt.ylabel('Runtime (seconds)')
plt.legend()
plt.show()
```
