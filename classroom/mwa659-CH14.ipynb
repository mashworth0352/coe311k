```python
import numpy as np
import matplotlib.pyplot as plt

N = 25
#I used 25 steps so there was clear differentiation between steps but was still relatively smooth
a = 0
b = 2*np.pi

h = (b - a) / (N - 1)

#Central Approx

x = np.linspace(a, b, N)
y = np.sin(x)
df_dx=np.cos(x)
df_dx_CA = (np.sin(x+h) - np.sin(x-h)) / (2 * h)


#Forward Approx

df_dx_FA = (np.sin(x + h) - y)/h


#Backwards Diff Approx

df_dx_BDA = (y - np.sin(x-h))/h

plt.plot(x, df_dx, 'o--', label='Exact Derivative (cos(x))')
plt.plot(x, df_dx_CA, label='Central Difference Approximation')
plt.plot(x, df_dx_FA, label='Forward Difference Approximation')
plt.plot(x, df_dx_BDA, label='Backwards Difference Approximation')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title("Approximating Derivative of f(x) = sin(x) using Various Approximations")
plt.legend()
plt.show()
```
