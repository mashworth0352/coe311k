```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,2*np.pi,50)
y = np.sin(x)
y2 = np.cos(x)
y3 = (np.sin(x))*4-(np.cos(x))*(2)
y4 = np.tan(x)
plt.figure()
plt.subplot(2,2,1)
plt.plot(x,y,'r')
plt.subplot(2,2,2)
plt.plot(x,y2,'b')
plt.subplot(2,2,3)
plt.plot(x,y3,'k')
plt.subplot(2,2,4)
plt.plot(x,y4,'g')
plt.show()
```
