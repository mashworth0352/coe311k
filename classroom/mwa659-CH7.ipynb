```python
import numpy as np

#Jacobi function:
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

#Gauss-Seidel function:
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

A = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

b = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
solution, iterations= jacobi(A, b, x0, tol, max_iterations)
solutiongs, iterationsgs= gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution for A Jacobi: {solution}")
print(f"Iterations: {iterations}")
print(f"Solution of A Gauss_Seidel: {solutiongs}")
print(f"Iterations: {iterationsgs}")

#-----------------------------------------------------------------------------------------------
A = np.array([[1, 1], [1, 1.0001]])
b = np.array([1, 1])
solution, iterations= jacobi(A, b, x0, tol, max_iterations)
solutiongs, iterationsgs= gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution for A  part 2 Jacobi: {solution}")
print(f"Iterations: {iterations}")
print(f"Solution of A part 2 Gauss_Seidel: {solutiongs}")
print(f"Iterations: {iterationsgs}")
#-----------------------------------------------------------------------------------------------
A = np.array([[1, 1], [1, 1.0001]])
b = np.array([1, 1.0001])
solution, iterations= jacobi(A, b, x0, tol, max_iterations)
solutiongs, iterationsgs= gauss_seidel(A, b, x0, tol, max_iterations)
print(f"Solution for A part 3 Jacobi: {solution}")
print(f"Iterations: {iterations}")
print(f"Solution of A part 3 Gauss_Seidel: {solutiongs}")
print(f"Iterations: {iterationsgs}")
```
