```python
import numpy as np
import matplotlib.pyplot as plt

def leibnizPi(n):
    estiPi=0
    counter=0
    for i in range(n):
        estiPi = estiPi + 4*((-1)**i)/(2*i+1)
        counter = counter + 1
        if (counter%100==0):
            print('Relative error for the', i + 1 , 'th term is',np.fabs(estiPi-np.pi)/np.pi)
    print('Final estimated pi value =',estiPi)
    return

leibnizPi(5000)

#The error between the calculated and actual values of pi decreases (very slowly)
#as each term in the summation is added.
```
