```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import random

#Part 1-----------------------------------------------------------------------------------------------------------

X = np.array([1,2,3,4,5])
Y = np.array([1.2,2.8,3.6,4.5,5.1])
A = np.column_stack([X, np.ones(len(X))])
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
y = m*X + c

print("Slope:", m)
print("Intercept:", c)

plt.figure()
plt.scatter(X,Y)
plt.plot(X,y)
plt.title("Part 1: Least Squares Fit")
plt.xlabel("X")
plt.ylabel("Y")

#Part 2------------------------------------------------------------------------------------------------------------

ind = random.permutation(len(X))
Xrand = X[ind]
Yrand = Y[ind]
A2 = np.column_stack([Xrand, np.ones(len(Xrand))])
m2, c2 = np.linalg.lstsq(A2, Yrand, rcond=None)[0]
y2 = m2*Xrand + c2

print("New slope:", m2)
print("New intercept:", c2)

plt.figure()
plt.scatter(Xrand,Yrand)
plt.plot(Xrand,y2)
plt.title("Part 2: Randomized Data")
plt.xlabel("X")
plt.ylabel("Y")

#The best fit line that is produced by randomizing the vectors is the same than the previous one.
#This suggests that the order the vectors are presented does not matter, but rather the vector values themselves.


#Part 3-------------------------------------------------------------------------------------------------------------

m_true = 2
b_true = 1

x3 = np.arange(0,1.01,0.01)
noise = random.rand(len(x3))
sign = random.choice([-1,1], len(x3))
noise = noise*sign
y3 = m_true*x3 + b_true + noise
A3 = np.column_stack([x3, np.ones(len(x3))])
m3, b3 = np.linalg.lstsq(A3, y3, rcond=None)[0]

print("Original line: y =", m_true,"x +", b_true)
print("Recovered line: y =", m3,"x +", b3)

y_pred = m3*x3 + b3

plt.figure()
plt.scatter(x3,y3)
plt.plot(x3,y_pred)
plt.title("Part 3: Noisy Data Fit")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
