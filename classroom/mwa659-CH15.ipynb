```python
import numpy as np
import matplotlib.pyplot as plt

h = np.arange(0.1,1.05,0.05)
x_start, x_end = 0, 2

def f(x, y):
    return -2 * y

allx = []
ally = []

for j in range(h.size):
    num_steps = int((x_end - x_start) / h[j])
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1
    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h[j] * f(x_values[i], y_values[i])
    
    allx.append(x_values)
    ally.append(y_values)

    plt.plot(x_values, y_values, label=("Euler's Method Approx. using step size " f"{h[j]:.2f}"))
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title("Approximation of dy/dx = -2y using Euler's Method")
    plt.legend()
    plt.show()    

xr, yr = allx[0], ally[0]

for j in range(1, h.size):
    
    xc, yc = allx[j], ally[j]
    error = np.abs(yc - np.interp(xc, xr, yr))

    print(f"\nh = {h[j]:.2f}")
    print("Error at each point:")
    print(error)
    print("Total accumulated error =", np.sum(error))
```
