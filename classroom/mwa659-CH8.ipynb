```python
import numpy as np

def diagonally_dominant(m):

    n = len(m)

    for i in range(n):

        diagonal = abs(m[i][i])
        othersum = 0

        for j in range(n):
            if j != i:
                othersum += abs(m[i][j])

        if diagonal < othersum:
            return False

    return True

#true example:

A = [[4,1,1],
     [2,7,1],
     [1,1,5]]

print("Is the matrix diagonally dominant?", diagonally_dominant(A))

#false example:

B = [[4,3,6],
     [9,7,0],
     [1,4,5]]

print("Is the matrix diagonally dominant?", diagonally_dominant(B))
```
