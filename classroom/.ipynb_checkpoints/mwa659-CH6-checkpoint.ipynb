```python
#This code answers both challenges 6a and 6b
import numpy as np
import time

#Jacobi function from class:
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

#Gauss-Seidel function:
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

#Challenge 6a-----------------------------------------------------------------
A = np.array([[10,-1,2,0,0],
            [-1,11,-1,3,0],
            [2,-1,10,-1,0],
            [0,3,-1,8,-2],
            [0,0,0,-2,9]])
b = np.array([14, 30, 26, 25, 37])
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100

start1 = time.perf_counter()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
end1 = time.perf_counter()
time_jacobi = end1 - start1

print(f"Solution for 6a Jacobi: {solution}")
print(f"Iterations: {iterations}")
print(f"Time: {time_jacobi:.6e} seconds")

Ags = np.array([[10,-1,2,0,0],
            [-1,11,-1,3,0],
            [2,-1,10,-1,0],
            [0,3,-1,8,-2],
            [0,0,0,-2,9]])
bgs = np.array([14, 30, 26, 25, 37])
x0gs = np.zeros_like(b)
tolgs = 1e-6
max_iterationsgs = 100

start2 = time.perf_counter()
solutiongs, iterationsgs = gauss_seidel(Ags, bgs, x0gs, tolgs, max_iterationsgs)
end2 = time.perf_counter()
time_gs = end2 - start2

print(f"Solution of 6a Gauss_Seidel: {solutiongs}")
print(f"Iterations: {iterationsgs}")
print(f"Time: {time_gs:.6e} seconds")

#Comparison of convergence in 6a--------------
#The Jacobi method converged after only 1 iteration, but the Gauss-Seidel took 2 iterations.
#Although the Jacobi method took one iteration, it can be seen that its speed was still slower than that of the G-S method.

#Challenge 6b-------------------------------------------------------------------
A2 = np.array([[1,2,3,0,0],
              [2,1,2,3,0],
              [3,2,1,2,3],
              [0,3,2,1,2],
              [0,0,3,2,1]])
b2 = np.array([14,22,33,26,22])
x02 = np.zeros_like(b2)
tol2 = 1e-6
max_iterations2 = 100

start3 = time.perf_counter()
solution2, iterations2 = jacobi(A2, b2, x02, tol2, max_iterations2)
end3 = time.perf_counter()
time_jacobi2 = end3 - start3

print(f"Solution for 6b Jacobi: {solution2}")
print(f"Iterations: {iterations2}")
print(f"Time: {time_jacobi2:.6e} seconds")

Ags2 = np.array([[1,2,3,0,0],
              [2,1,2,3,0],
              [3,2,1,2,3],
              [0,3,2,1,2],
              [0,0,3,2,1]])
bgs2 = np.array([14,22,33,26,22])
x0gs2 = np.zeros_like(b2)
tolgs2 = 1e-6
max_iterationsgs2 = 100

start4 = time.perf_counter()
solutiongs2, iterationsgs2 = gauss_seidel(Ags2, bgs2, x0gs2, tolgs2, max_iterationsgs2)
end4 = time.perf_counter()
time_gs2 = end4 - start4

print(f"Solution of 6b Gauss_Seidel: {solutiongs2}")
print(f"Iterations: {iterationsgs2}")
print(f"Time: {time_gs2:.6e} seconds")

#Comparison of convergence in 6b--------------
#Due to the structure of the martix (A2/Ags2) not being diagonally-dominant, both methods diverged from a defined answer.
#Regardless, the Jacobi method still took longer to compute the max iterations set at 100 than the G-S method.
```
