```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import default_rng
def montecarlo_pi(samples):
    x = 0
    y = 0
    counter = 0
    for s in range (samples):
    
        x = rng.random(1)*2-1
        y = rng.random(1)*2-1
        #random number between -1 and 1
        if (x**2 + y**2 < 1):
            counter = counter + 1
    
    
    return 4.0 * counter / samples

rng = default_rng()
abs_error= np.zeros(100)
rel_error = np.zeros(100)
iteration = np.arange(1,101,1)

for i in range (100):
    abs_error[i] = np.abs(np.pi - montecarlo_pi(1000))
    rel_error[i] = np.abs((np.pi - montecarlo_pi(1000))/np.pi)

plt.figure()
plt.plot(iteration,abs_error, 'r')
plt.plot(iteration,rel_error,'b')
plt.xlabel('Iteration')
plt.ylabel('Pi Error')
plt.title('Montecarlo Pi Errors vs Calculated Iterations')
plt.legend(['Absolute Error','Relative Error'])
plt.grid()
plt.show()
    
```
