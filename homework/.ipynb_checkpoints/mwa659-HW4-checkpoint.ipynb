```python
import numpy as np
import matplotlib.pyplot as plt

#Jacobi function:
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    errors = []
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        errors.append(np.linalg.norm(x_new - x, ord=np.inf))
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1, errors
        x = x_new
    return x, max_iterations, errors

#Gauss-Seidel function:
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    errors = []
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        errors.append(np.linalg.norm(x_new - x, ord=np.inf))
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1, errors
        x = x_new
    return x, max_iterations, errors

A = np.array([
    [3,   -0.1, -0.2],
    [0.1,  7,   -0.3],
    [0.3, -0.2, 10]
    ])
b = np.array([7.85, -19.3, 71.4])
x0 = np.zeros_like(b)
tol = np.array([1e-3,1e-6,1e-9])
max_iterations = 100
jacobi_errors_all = []
gs_errors_all = []

for i in range(3):
    solution, iterations, errors = jacobi(A, b, x0, tol[i], max_iterations)
    solutiongs, iterationsgs, errorsgs = gauss_seidel(A, b, x0, tol[i], max_iterations)
    jacobi_errors_all.append(errors)
    print(f"{jacobi_errors_all} : {errors}" )
    gs_errors_all.append(errorsgs)

    print(f"Solution for A with tolerance {tol[i]} using Jacobi:", solution)
    print(f"Iterations: {iterations}")
    print(f"Solution of A with tolerance {tol[i]} using Gauss-Seidel:", solutiongs)
    print(f"Iterations: {iterationsgs}")
    
#COMPARISON OF ITERATIONS BETWEEN EACH TYPE----------------------------------------------------------

#The stricter tolerances require more iterations for both methods of solving.
#The Jacobi method generally requires more iterations than the Gauss_Seidel,
#but the lower tolerance (1e-3) used the same amount of both tolerances for both methods.

#ERROR ANALYSIS--------------------------------------------------------------------------------------

for i in range(3):
    plt.figure()

    plt.plot(
        range(1, len(jacobi_errors_all[i]) + 1),
        jacobi_errors_all[i],
        label=f'Jacobi tol={tol[i]}'
    )

    plt.plot(
        range(1, len(gs_errors_all[i]) + 1),
        gs_errors_all[i],
        linestyle='--',
        label=f'GS tol={tol[i]}'
    )

    plt.xlabel('Iterations')
    plt.ylabel('Absolute Error')
    plt.title(f'Convergence Comparison (tol = {tol[i]})')
    plt.yscale('log')
    plt.grid()
    plt.legend()
    plt.show()

#TOLERANCE DISCUSSION---------------------------------------------------------------------------------

#As previously discussed, a stricter tolerance (i.e. 1e-9 is stricter than 1e-6)
#requires more iterations for convergence than looser tolerances.
#This is due to the fact that a smaller error margin requires more values to be included in both methods.
```
