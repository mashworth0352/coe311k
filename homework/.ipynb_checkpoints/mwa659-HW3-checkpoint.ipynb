```python
import numpy as np
import math
import matplotlib.pyplot as plt

def sinTaylor(n,terms):
    sinT = 0
    termError = 0
    for i in range (terms):
        sinT = sinT + ((-1)**i)*(n**(2*i+1))/math.factorial(2*i+1)
    termError = np.fabs(np.sin(n)-sinT)
    return termError

nValues = np.arange(0,10.1,0.1)
oneTerm = np.zeros(nValues.size)
twoTerm = np.zeros(nValues.size)
threeTerm = np.zeros(nValues.size)
fourTerm = np.zeros(nValues.size)

for j in range (nValues.size):
    oneTerm[j] = sinTaylor(nValues[j],1)
    twoTerm[j] = sinTaylor(nValues[j],2)
    threeTerm[j] = sinTaylor(nValues[j],3)
    fourTerm[j] = sinTaylor(nValues[j],4)

plt.figure()
plt.title('Abs Error of Taylor Series sin(x) vs Tested Values')
plt.plot(nValues,oneTerm,'r')
plt.plot(nValues,twoTerm,'b')
plt.plot(nValues,threeTerm,'g')
plt.plot(nValues,fourTerm,'k')
plt.xlabel('Tested Value')
plt.ylabel('Absolute Error')
plt.legend(['One Term','Two Terms','Three Terms','Four Terms'])
plt.grid()
plt.show()

#The sinx taylor series for all terms initially has minimal error between the actual function and each other at lower x values.
#However, as x approaches 10, the error does not decrease uniformly;
#in fact, the errors for all terms gradually skyrocket at higher x values.
#The one term series sees the least amount of error climb, while the error increases more with each additional term.
```
