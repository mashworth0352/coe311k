```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import default_rng
def montecarlo_pi(samples):
    x = 0
    y = 0
    counter = 0
    for s in range (samples):
    
        x = rng.random(1)*2-1
        y = rng.random(1)*2-1
        #random number between -1 and 1
        if (x**2 + y**2 < 1):
            counter = counter + 1
    
    
    return 4.0 * counter / samples

rng = default_rng()
sample_count = np.arange(10,10000,20)
z = np.zeros(np.size(sample_count))

for i in range (np.size(sample_count)):
    z[i] = np.round(np.abs(montecarlo_pi(sample_count[i])-np.pi), 15)
    #round 15 decimal places

plt.figure()
plt.plot(sample_count,z)
plt.xlabel('Sample Size')
plt.ylabel('Pi Error')
plt.title('Montecarlo Pi Error vs Sample Size')
plt.grid()
plt.show()

    

```
